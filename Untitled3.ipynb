```python
import sys
import numpy as np, pandas as pd
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
path = sys.argv[1] if len(sys.argv) > 1 else "household_power_consumption.txt"
```

```python
# Fallback to default if path was set to jupyter's internal '-f' flag
correct_path = "household_power_consumption.txt" if path.startswith("-") else path

raw = pd.read_csv(correct_path, sep=";", na_values=["?"], low_memory=False, usecols=["Date", "Global_active_power"])
raw["Date"] = pd.to_datetime(raw["Date"], format="%d/%m/%Y")
raw["Global_active_power"] = pd.to_numeric(raw["Global_active_power"], errors="coerce")
print("Rows:", len(raw), "| missing:", raw["Global_active_power"].isna().sum())
```

```python
daily = raw.groupby("Date")["Global_active_power"].mean().asfreq("D")
print("Days:", len(daily), "| days with no data:", daily.isna().sum())
daily = daily.interpolate().rename("power")
daily.plot(figsize=(11, 4), title="Daily average power (kW)")
plt.tight_layout(); plt.savefig("trend.png", dpi=120); plt.close()
```

```python
df = pd.DataFrame({"power": daily})
df["lag_1"] = df["power"].shift(1)
df["lag_7"] = df["power"].shift(7)
df["roll_7"] = df["power"].shift(1).rolling(7).mean()
df["dow"] = df.index.dayofweek
df["month"] = df.index.month
df = df.dropna()
```

```python
cut = int(len(df) * 0.8)
train, test = df.iloc[:cut], df.iloc[cut:]
feats = ["lag_1", "lag_7", "roll_7", "dow", "month"]
```

```python
preds = {"Naive (yesterday)": test["lag_1"].values}
for name, m in {"Linear Regression": LinearRegression(),
                "Random Forest": RandomForestRegressor(n_estimators=200, min_samples_leaf=3, random_state=42)}.items():
    m.fit(train[feats], train["power"])
    preds[name] = m.predict(test[feats])
```

```python
y = test["power"].values
rows = []
for n, p in preds.items():
    rows.append([n, mean_absolute_error(y, p), np.sqrt(mean_squared_error(y, p)),
                 r2_score(y, p), np.mean(np.abs((y - p) / y)) * 100])
res = pd.DataFrame(rows, columns=["Model", "MAE", "RMSE", "R2", "MAPE %"]).round(3)
print(res.to_string(index=False)); res.to_csv("power_results.csv", index=False)
```

```python
plt.figure(figsize=(12, 5))
plt.plot(train.index[-120:], train["power"].tail(120), label="Train (last 120 days)")
plt.plot(test.index, y, label="Actual", color="black")
plt.plot(test.index, preds["Random Forest"], label="Random Forest", alpha=.8)
plt.plot(test.index, preds["Linear Regression"], label="Linear Regression", alpha=.8)
plt.title("Actual vs Predicted daily power (kW)"); plt.legend(); plt.tight_layout()
plt.savefig("power_actual_vs_predicted.png", dpi=120)
```
